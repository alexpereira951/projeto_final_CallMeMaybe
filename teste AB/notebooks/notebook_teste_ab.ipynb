# Bibliotecas e Funções usadas

## Bibliotecas:

```python
import pandas as pd
import plotly.express as px
from statsmodels.stats.proportion import proportions_ztest
```

## Funções:

#### Função ***teste_z***: Essa função executa o teste z para a igualdade de proporções

```python
def teste_z(grupo_1_evento: int, grupo_2_evento:int , total_grupo_1: int, total_grupo_2: int, alpha=0.01):
    '''
    Essa função executa o teste z para a igualdade de proporções

    INPUTS:
    grupo_1_evento: Quantidade de Usuários do grupo 1 que realizaram o evento (int)
    grupo_2_evento: Quantidade de Usuários do grupo 2 que realizaram o evento (int)
    total_grupo_1: Total de usuários do grupo 1 (int)
    total_grupo_2: Total de usuários do grupo 2 (int)
    '''
    
    # Teste Z
    z_stat, p_value = proportions_ztest([grupo_1_evento, grupo_2_evento], [total_grupo_1, total_grupo_2])
    print(f"Z-statistic: {z_stat:.4f}")
    print(f"P-value: {p_value:.4f}")
    print(f"Alpha: {alpha}")
    
    if p_value < alpha:
        return  '❌Rejeitamos a hipótese nula: A conversão entre os grupos foi DIFERENTE'
    else:
        return  '✅Aceitamos a hipótese nula: A conversão entre os grupos foi IGUAL, logo o teste não trouxe melhorias.'
```

## 2.2. O número de eventos por usuário é distribuído igualmente entre as amostras?

```python
# Calculando o numero de eventos por usuario para cada grupo:
events_per_group = (
    to_funil.groupby(["group", "user_id"])["event_name"].count().reset_index()
)

events_A = events_per_group[events_per_group["group"] == "A"]
events_B = events_per_group[events_per_group["group"] == "B"]

# Comparando a distribuição do numero de eventos por usuario de cada grupo:
fig = px.histogram(
    events_per_group,
    x="event_name",
    color="group",
    facet_row="group",
    barmode="overlay",
    labels={
        "event_name": "Número de Eventos por Usuário",
        "group": "Grupo",
        "count": "Contagem de Usuários",
    },
    title="Distribuição do Número de Eventos por Usuário por Grupo",
    height=600,
)

fig.update_traces(opacity=0.75)

color_map = {"A": "#636EFA", "B": "#EF553B"}

for g in events_per_group["group"].unique():
    df_group = events_per_group[events_per_group["group"] == g]
    mean_val = df_group["event_name"].mean()
    median_val = df_group["event_name"].median()
    color = color_map.get(g, "black")

    
    row_idx = 2 if g == "A" else 1

    fig.add_vline(
        x=mean_val,
        line_dash="dash",
        line_color=color,
        annotation_text=f"Média: {mean_val:.2f}",
        annotation_position="top right",
        row=row_idx,
        col=1,
    )

    fig.add_vline(
        x=median_val,
        line_dash="dot",
        line_color=color,
        annotation_text=f"Mediana: {median_val:.2f}",
        annotation_position="top left",
        row=row_idx,
        col=1,
    )

fig.for_each_annotation(
    lambda a: a.update(text=a.text.split("=")[-1]) if "group=" in a.text else None
)

fig.show()
```

conclusão:

A análise dos histogramas revela uma discrepância crítica no comportamento e no engajamento dos usuários entre as duas amostras, mostrando que os grupos não estão distribuídos igualmente:

Dominância de Baixo Engajamento no Grupo B: O Grupo B apresenta uma concentração massiva e desproporcional de usuários na faixa mínima de eventos (representada pela primeira barra do gráfico). Essa enorme quantidade de usuários que realizam apenas 1 evento e abandonam a plataforma arrasta a mediana do Grupo B para o valor mínimo (0 ou 1, dependendo do bin) e derruba a média para apenas 1.43 eventos por usuário.

Maior Retenção e Atividade no Grupo A: Em contrapartida, o Grupo A demonstra uma distribuição muito mais saudável e espalhada ao longo da cauda. Os usuários do Grupo A não apenas passam da primeira etapa, como continuam interagindo recorrentemente, o que resulta em uma média de 5.10 eventos por usuário e uma mediana de 4.00.

Em resumo: O Grupo B falhou gravemente em reter os usuários logo após o primeiro contato, enquanto o Grupo A garantiu uma jornada muito mais profunda e engajada. Essa diferença nas métricas centrais indica que o fator que diferencia o Grupo B (seja uma nova feature, página ou campanha) pode estar gerando um impacto negativo na retenção inicial.

- 🎄 Campanha de marketing sobreposta (Christmas & New Year Promo na EU): Período sazonal (dezembro/janeiro — comportamento atípico dos usuários)

  Se há uma campanha de marketing ativa durante o teste, não podemos saber se os resultados são causados pela mudança testada ou pela campanha.

- ⚖️ Grupos muito desiguais (~75% no grupo A vs ~25% no grupo B)

  Se os grupos estão muito desiguais em tamanho, as comparações entre eles são justas, gerando mais certeza para um grupo em comparação com o outro.

- 👥 Poucos participantes (2788 vs 6000 esperados — menos da metade!)

  Se o número de participantes está abaixo da metade do esperado, não conseguimos detectar diferenças reais com confiança.
