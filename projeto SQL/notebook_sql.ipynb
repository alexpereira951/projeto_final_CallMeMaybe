# 2. Encontrando o número de livros lançados depois de 1 de janeiro de 2000.

```python
# Precisamos escrever uma query em SQL para contar o número de livros lançados depois de '2000-01-01' e armazenar o resultado na coluna 'livros_apos_2000':
query_count = ''' 
SELECT 
    COUNT(book_id) AS livros_apos_2000
FROM 
    books
WHERE 
    CAST(publication_date AS date) > '2000-01-01';
'''

livros_apos_2000 = conexao(query_count)

livros_apos_2000
```

conclusão: 

Temos 819 livros lançados após 1 de janeiro de 2000

# 3. Encontrando o número de avaliações e a classificação média para cada livro.

```python
query_rating_reviews = ''' 
SELECT 
    books.title AS livro,
    AVG(ratings.rating) AS avg_rating,
    COUNT(reviews.review_id) AS count_reviews
FROM 
    books
INNER JOIN
    ratings 
    ON ratings.book_id = books.book_id 
INNER JOIN
    reviews 
    ON reviews.book_id = ratings.book_id
GROUP BY
    books.title
ORDER BY
    avg_rating DESC,
    count_reviews DESC
'''

rating_and_reviews = conexao(query_rating_reviews)

rating_and_reviews.head()
```

conclusão:

Os top 3 títulos com maior número de reviews e avaliação média são respectivamente:

- 1°: A Dirty Job (Grim Reaper #1)
- 2°: School's Out—Forever (Maximum Ride #2)
- 3°: Moneyball: The Art of Winning an Unfair Game

Recomendações:

Criar campanhas de venda casada (cross-selling) e destaque esses três títulos na home, pois eles são seus maiores ímãs de conversão e tráfego orgânico.

# 4. Identificando a editora que lançou o maior número de livros com mais de 50 páginas 

```python
query_editoras = ''' 
SELECT 
    publishers.publisher as editora,
    COUNT(books.book_id) AS livros_com_mais_de_50_pg
FROM 
    books
INNER JOIN 
    publishers
    ON publishers.publisher_id = books.publisher_id
WHERE
    books.num_pages > 50
GROUP BY 
    publishers.publisher
ORDER BY
    livros_com_mais_de_50_pg DESC
'''

editoras_50 = conexao(query_editoras)
editoras_50.head()
```

conclusão:

O top 3 editoras que lançaram o maior número de livros com mais de 50 páginas é:

- 1°: Penguin Books
- 2°: Vintage
- 3°: Grand Central Publishing

Recomendações: 

Estreitar o relacionamento comercial e negocie margens de lucro agressivas com Penguin, Vintage e Grand Central, já que elas garantem a volumetria do ecossistema.

# 5. Identificando o autor com a média mais alta de classificação de livros: Olharemos apenas para livros com pelo menos 50 classificações.

```python
query_editoras = ''' 
SELECT 
    authors.author AS autor,
    AVG(books_50.avg_book_rating) AS avg_rating
FROM (-- Subquery selecionando apenas os livros com pelo menos 50 classificações
    SELECT 
        book_id,
        AVG(rating) AS avg_book_rating
    FROM ratings
    GROUP BY book_id
    HAVING COUNT(rating) >= 50
) AS books_50
INNER JOIN books 
    ON books.book_id = books_50.book_id
INNER JOIN authors 
    ON authors.author_id = books.author_id
GROUP BY 
    authors.author
ORDER BY 
    avg_rating DESC
'''

autores_50 = conexao(query_editoras)
autores_50.head()
```

conclusão:

O pódio dos autores com a média mais alta de classificação de livros lançados, para livros com pelo menos 50 classificações é:

- 1°: J.K. Rowling/Mary GrandPré
- 2°: Markus Zusak/Cao Xuân Việt Khương
- 3°: J.R.R. Tolkien

Recomendações:

Garantir exclusividade ou campanhas de pré-venda para lançamentos desses autores (ou coautores), pois a base de fãs deles compra de olhos fechados.

# 6. Encontrando o número médio de avaliações entre usuários que classificaram mais do que 50 livros

```python
query_usuarios = '''
SELECT 
    AVG(reviews_50.quantidade_de_avaliacoes) AS avg_reviews
FROM (-- Subquery para selecionar apenas os usuários que clasificaram mais de 50 livros
    SELECT
        username,
        COUNT(rating_id) As quantidade_de_classificacoes
    FROM 
        ratings
    GROUP BY 
        username
    HAVING 
        COUNT(rating_id) > 50
    ) AS ratings_50
INNER JOIN 
    (-- Subquery para contar a quantidade de avaliações por usuário
    SELECT
        username,
        COUNT(review_id) As quantidade_de_avaliacoes
    FROM 
        reviews
    GROUP BY 
        username
    ) AS reviews_50
    ON reviews_50.username = ratings_50.username
'''

# Executando a consulta
resultado_usuarios = conexao(query_usuarios)
resultado_usuarios
```

conclusão: 

Os usuários mais ativos na plataforma (Aqueles que classificam mais de 50 livros) possuem um número médio de 24 avaliações.

Recomendações:

Criar um programa de fidelidade VIP ou gamificação que recompense o usuário ao atingir a marca de 24 avaliações, retendo os clientes que mais geram prova social gratuita para a plataforma.
